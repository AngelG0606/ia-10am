## 1. Objetivo
Desarrollar criterio analítico para diferenciar entre:
- Obtención pasiva de datos (búsqueda)
- Creación activa de datos (instrumentación)
- Análisis Exploratorio de Datos (EDA)



## 2. Fase 1: Análisis de Casos

### Escenario A: Búsqueda e Integración
Datos provenientes de fuentes externas. Problemas: limpieza, integración y consistencia.

### Escenario B: Creación e Instrumentación
Datos generados mediante sensores. Problemas: ruido, calibración y adquisición.


```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

np.random.seed(42)
tiempo = pd.date_range(start='2026-04-15', periods=500, freq='min')

valores = np.linspace(10, 50, 500) + np.sin(np.linspace(0, 20, 500))*5 + np.random.normal(0, 2, 500)

df = pd.DataFrame({'Timestamp': tiempo, 'Lectura': valores})
df.set_index('Timestamp', inplace=True)

plt.figure(figsize=(10,4))
plt.plot(df.index, df['Lectura'])
plt.title("Serie de Tiempo")
plt.grid()
plt.show()

sns.histplot(df['Lectura'], kde=True)
plt.title("Histograma")
plt.show()

df.describe()
```

## 3. Inyección de anomalías
Se introducen valores atípicos para analizar su impacto.


## Histograma de las lecturas

![Histograma](/mnt/data/bf572219-f1fd-49d3-aff2-617cb0e7e837.png)

```python

valores[100] = 120
valores[250] = -30
valores[300:310] = np.nan

df['Lectura'] = valores

plt.figure(figsize=(10,4))
plt.plot(df.index, df['Lectura'])
plt.title("Serie con anomalías")
plt.grid()
plt.show()

sns.histplot(df['Lectura'], kde=True)
plt.title("Histograma con anomalías")
plt.show()

df.describe()
```
